```r
# Parameters
param_seagrass_species <- list("Cymodocea nodosa", "Halophila stipulacea", "Posidonia oceanica", "Zostera marina", "Zostera noltei", "Zostera marina and Cymodocea nodosa", "Zostera marina and Zostera noltei")
param_latitude <- "56.0953"
param_longitude <- "14.2785"
```

```r
# Seagrass species setter
seagrass_species <- list("Cymodocea nodosa", "Halophila stipulacea", "Posidonia oceanica", "Zostera marina", "Zostera noltei", "Zostera marina and Cymodocea nodosa", "Zostera marina and Zostera noltei")
```

```r
# Parameter validator

validate_items_are_in_list <- function(items, list_) {
  is_present <- items %in% list_
  if (!all(is_present)) {
    missing_items <- items[!is_present]
    stop(
      "The following item(s) are missing from the available list: ", 
      paste(missing_items, collapse = ", ")
    )
  } else {
    message("Success: All required items were found in the available list.")
  }
}

validate_geographic_coordinate <- function(coordinate_string) {  
  pattern <- "^\\d{2}\\.\\d{4}$"
  is_valid <- grepl(pattern, coordinate_string)
  if (!is_valid) {
      stop(
          "The coordinate: ", coordinate_string, " does not follow the required format: ", pattern, " e.g. 56.0953"
          )
      }
  else {
      message(coordinate_string, " is valid geographic coordinate input.")
      }
}

validate_items_are_in_list(param_seagrass_species, seagrass_species)
validate_geographic_coordinate(param_latitude)
validate_geographic_coordinate(param_longitude)
```
